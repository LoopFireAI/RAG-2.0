## **Ingesting Test**

This is a test to see if we are able to ingest .ppt or .pptx documents from Google Drive.

```python
from langchain_google_community import GoogleDriveLoader
from langchain_core.documents import Document
from typing import List
import os
```

```python
import os
from langchain_google_community import GoogleDriveLoader

# Replace with your actual file_id
folder_id = '1SBr_c2NwCuXsNZsjau37uxxq2qRQH1F6'

# These environment variables should point to your credentials and token files
credentials_path = os.getenv("GOOGLE_CREDENTIALS_PATH")
token_path = os.getenv("GOOGLE_TOKEN_PATH")

if not credentials_path:
    raise ValueError("GOOGLE_CREDENTIALS_PATH environment variable is not set.")
if not token_path:
    raise ValueError("GOOGLE_TOKEN_PATH environment variable is not set.")
if not os.path.exists(credentials_path):
    raise FileNotFoundError(f"Credentials file not found at: {credentials_path}.")
if not os.path.exists(token_path):
    raise FileNotFoundError(f"Token file not found at: {token_path}.")

# Initialize the loader for a single file
loader = GoogleDriveLoader(
    folder_id=folder_id, 
    credentials_path=credentials_path,
    token_path=token_path,
)

docs = loader.load()
print(len(docs))
for doc in docs:
    print("---")
    print(doc.metadata)
    print(doc.page_content[:500])  # Print first 500 chars for inspection
```
